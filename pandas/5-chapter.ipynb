```python
import pandas as pd

df = pd.read_csv('pandas_orders.csv', parse_dates=['order_date', 'signup_date'])
df
```

```python
df['rank'] = df.groupby("user_id")["order_date"].rank(ascending=True)
```

```python
df['lag'] = pd.to_datetime(df.groupby("user_id")["order_date"].shift(1, fill_value=0))
```

```python
df['delta'] = df['order_date'] - df['lag']
```

```python
df['revenue'] = df['price'] * df['quantity']
df['cumsum'] = df.groupby("user_id")["revenue"].cumsum()
```

```python
df['rank_sum'] = df[df['status']=='completed'].groupby("city")["revenue"].rank(ascending=True)
```

```python
df['count'] = df.groupby("user_id")['user_id'].transform('count')
```

```python
df['sum'] = df.groupby("user_id")["revenue"].transform('sum')
df['part'] = df['revenue'] / df['sum']
```
